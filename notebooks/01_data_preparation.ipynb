```python
def recommend_solution(new_ticket, top_k=10):

    # 1. Transformer le nouveau ticket en embedding
    new_embedding = model.encode([new_ticket])

    # 2. Calculer la similarité avec tous les tickets historiques
    similarities = cosine_similarity(
        new_embedding,
        embeddings
    )[0]

    # 3. Prendre les top_k tickets les plus similaires
    top_indices = similarities.argsort()[-top_k:][::-1]

    # 4. Récupérer les tickets
    candidates = local_2025.iloc[top_indices].copy()

    # 5. Ajouter la similarité
    candidates["Similarity"] = similarities[top_indices]

    # 6. Vérifier si la solution est recommandable
    candidates["recommendable_solution"] = (
        candidates["Solution"]
        .apply(is_recommendable_solution)
    )

    # 7. Garder uniquement les solutions recommandables
    recommended = candidates[
        candidates["recommendable_solution"]
    ].copy()

    # 8. Si aucune solution exploitable n'est trouvée
    if recommended.empty:
        return None, candidates

    # 9. Prendre le ticket avec la meilleure similarité
    best = recommended.sort_values(
        "Similarity",
        ascending=False
    ).iloc[0]

    return best, candidates
```

```python

def classify_solution(solution):
    solution = str(solution).strip().lower()

    generic_solutions = {
        "resolved",
        "resolved.",
        "resolved,",
        "solved",
        "solved.",
        "done",
        "done.",
        "fixed",
        "fixed.",
        "closed",
        "granted",
        "necessary done",
        "necessary done.",
        "necessaey done.",
        "issue fixed",
        "issue solved",
        "problem solved"
    }

    administrative_keywords = [
        "duplicate",
        "duplicated",
        "close the ticket",
        "ticket to avoid",
        "no response from user",
        "regional ticket opened",
        "new ticket has been raised",
        "a new ticket has been raised",
        "data provided"
    ]

    # 1. Solution exactement générique
    if solution in generic_solutions:
        return "GENERIC"

    # 2. Solution administrative
    for keyword in administrative_keywords:
        if keyword in solution:
            return "ADMINISTRATIVE"

    # 3. Tout le reste est considéré comme actionnable
    return "ACTIONABLE"
```

```python

def solution_match(recommended_solution, real_solution):
    """
    Vérifie si deux solutions sont suffisamment similaires.
    """
    recommended_solution = str(recommended_solution).strip().lower()
    real_solution = str(real_solution).strip().lower()

    # Correspondance exacte
    if recommended_solution == real_solution:
        return True

    # Une solution contient l'autre
    if (
            recommended_solution in real_solution
            or real_solution in recommended_solution
    ):
        return True

    return False
```

```python

def evaluate_recommendations(test_df):
    results = []

    for _, row in test_df.iterrows():

        ticket = row["Short Description"]
        real_solution = row["Solution"]

        recommendations = get_solution_recommendations(
            ticket,
            top_k=50,
            n_recommendations=5
        )

        recommended_solutions = (
            recommendations["Solution"]
            .tolist()
        )

        top1 = False
        top3 = False
        top5 = False

        if len(recommended_solutions) >= 1:
            top1 = solution_match(
                recommended_solutions[0],
                real_solution
            )

        if len(recommended_solutions) >= 3:
            top3 = any(
                solution_match(
                    solution,
                    real_solution
                )
                for solution in recommended_solutions[:3]
            )

        if len(recommended_solutions) >= 5:
            top5 = any(
                solution_match(
                    solution,
                    real_solution
                )
                for solution in recommended_solutions[:5]
            )

        results.append({
            "Number": row["Number"],
            "Real_Solution": real_solution,
            "Top1": top1,
            "Top3": top3,
            "Top5": top5
        })

    return pd.DataFrame(results)
```

```python
def search_best_solutions(query, top_k=5):
    """
    Recherche les meilleures solutions pour une nouvelle demande.
    Priorise les tickets escaladés et les solutions ACTIONABLE.
    """

    # Embedding de la requête
    query_embedding = model.encode(
        [query],
        normalize_embeddings=True
    )[0]

    # Normalisation des embeddings
    ticket_embeddings_norm = embeddings / np.linalg.norm(
        embeddings,
        axis=1,
        keepdims=True
    )

    # Similarité cosinus
    similarities = np.dot(
        ticket_embeddings_norm,
        query_embedding
    )

    # Copie de la base
    results = knowledge_base.copy()

    results["Similarity"] = similarities

    # Score final
    results["Final_Score"] = results["Similarity"]

    # Bonus pour les tickets escaladés
    results.loc[
        results["source"] == "ESCALATED",
        "Final_Score"
    ] += 0.05

    # Bonus supplémentaire pour les solutions actionnables
    results.loc[
        results["solution_type"] == "ACTIONABLE",
        "Final_Score"
    ] += 0.03

    # Classement
    results = results.sort_values(
        "Final_Score",
        ascending=False
    )

    return results[
        [
            "Number",
            "Short Description",
            "Solution",
            "solution_type",
            "source",
            "Similarity",
            "Final_Score"
        ]
    ].head(top_k)
```

```python
# ==========================================
# MOTEUR DE RECOMMANDATION DE SOLUTION
# ==========================================

def recommend_solutions(
    query,
    top_k_tickets=20,
    top_k_solutions=5
):
    """
    Recherche les incidents historiques les plus proches
    et retourne les solutions ACTIONABLE les plus pertinentes.
    """

    # --------------------------------------
    # 1. Embedding du nouveau ticket
    # --------------------------------------

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True
    )[0]

    # --------------------------------------
    # 2. Normalisation des embeddings tickets
    # --------------------------------------

    ticket_embeddings_norm = (
        embeddings /
        np.linalg.norm(
            embeddings,
            axis=1,
            keepdims=True
        )
    )

    # --------------------------------------
    # 3. Similarité nouveau ticket /
    #    tickets historiques
    # --------------------------------------

    ticket_similarities = (
        ticket_embeddings_norm @ query_embedding
    )

    # --------------------------------------
    # 4. Récupération des tickets similaires
    # --------------------------------------

    top_indices = np.argsort(
        ticket_similarities
    )[::-1][:top_k_tickets]

    candidates = knowledge_base.iloc[
        top_indices
    ].copy()

    candidates["Ticket_Similarity"] = (
        ticket_similarities[top_indices]
    )

    # --------------------------------------
    # 5. Garder les solutions exploitables
    # --------------------------------------

    actionable = candidates[
        candidates["solution_type"] == "ACTIONABLE"
    ].copy()

    # --------------------------------------
    # 6. Embedding de la solution du ticket
    #    par rapport à la solution des candidats
    # --------------------------------------

    solution_similarities = []

    for idx in actionable.index:

        solution_embedding = solution_embeddings[idx]

        similarity = np.dot(
            solution_embedding,
            query_embedding
        )

        solution_similarities.append(
            similarity
        )

    actionable["Solution_Similarity"] = (
        solution_similarities
    )

    # --------------------------------------
    # 7. Affichage des meilleures solutions
    # --------------------------------------

    actionable = actionable.sort_values(
        "Ticket_Similarity",
        ascending=False
    )

    return actionable[
        [
            "Number",
            "Short Description",
            "Solution",
            "Ticket_Similarity",
            "Solution_Similarity",
            "source",
            "solution_type"
        ]
    ].head(top_k_solutions)
```

```python
# ==========================================
# MOTEUR DE RECOMMANDATION V2
# ==========================================

def recommend_solutions_v2(
    query,
    top_k_tickets=50,
    top_k_solutions=5,
    ticket_weight=0.70,
    solution_weight=0.30
):
    """
    Recherche les tickets historiques similaires,
    puis classe les solutions ACTIONABLE selon
    un score combinant :

        70% : similarité du ticket
        30% : similarité de la solution
    """

    # --------------------------------------
    # 1. Embedding du nouveau ticket
    # --------------------------------------

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True
    )[0]

    # --------------------------------------
    # 2. Normalisation des embeddings tickets
    # --------------------------------------

    ticket_embeddings_norm = (
        embeddings /
        np.linalg.norm(
            embeddings,
            axis=1,
            keepdims=True
        )
    )

    # --------------------------------------
    # 3. Similarité nouveau ticket /
    #    tickets historiques
    # --------------------------------------

    ticket_similarities = (
        ticket_embeddings_norm @ query_embedding
    )

    # --------------------------------------
    # 4. Récupération des tickets similaires
    # --------------------------------------

    top_indices = np.argsort(
        ticket_similarities
    )[::-1][:top_k_tickets]

    candidates = knowledge_base.iloc[
        top_indices
    ].copy()

    candidates["Ticket_Similarity"] = (
        ticket_similarities[top_indices]
    )

    # --------------------------------------
    # 5. Garder uniquement ACTIONABLE
    # --------------------------------------

    actionable = candidates[
        candidates["solution_type"] == "ACTIONABLE"
    ].copy()

    # --------------------------------------
    # 6. Similarité entre la solution
    #    historique et le nouveau ticket
    # --------------------------------------

    solution_similarities = []

    for idx in actionable.index:

        solution_embedding = solution_embeddings[idx]

        similarity = np.dot(
            solution_embedding,
            query_embedding
        )

        solution_similarities.append(
            similarity
        )

    actionable["Solution_Similarity"] = (
        solution_similarities
    )

    # --------------------------------------
    # 7. SCORE FINAL
    # --------------------------------------

    actionable["Final_Score"] = (
        ticket_weight *
        actionable["Ticket_Similarity"]
        +
        solution_weight *
        actionable["Solution_Similarity"]
    )

    # --------------------------------------
    # 8. Trier par score final
    # --------------------------------------

    actionable = actionable.sort_values(
        "Final_Score",
        ascending=False
    )

    # --------------------------------------
    # 9. Retourner les meilleures solutions
    # --------------------------------------

    return actionable[
        [
            "Number",
            "Short Description",
            "Solution",
            "Ticket_Similarity",
            "Solution_Similarity",
            "Final_Score",
            "source",
            "solution_type"
        ]
    ].head(top_k_solutions)
```

```python
# ==========================================
# COMPARAISON DES SOLUTIONS HISTORIQUES
# ==========================================

from sklearn.metrics.pairwise import cosine_similarity
import numpy as np


def compare_historical_solutions(
    query,
    top_k_tickets=20,
    min_ticket_similarity=0.50
):
    """
    1. Recherche les tickets historiques similaires au nouveau ticket.
    2. Garde uniquement les solutions ACTIONABLE.
    3. Compare les solutions ACTIONABLE entre elles.
    """

    # --------------------------------------
    # 1. Embedding du nouveau ticket
    # --------------------------------------

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True
    )[0]

    # --------------------------------------
    # 2. Normalisation des embeddings tickets
    # --------------------------------------

    ticket_embeddings_norm = (
        embeddings /
        np.linalg.norm(
            embeddings,
            axis=1,
            keepdims=True
        )
    )

    # --------------------------------------
    # 3. Similarité nouveau ticket /
    #    historique
    # --------------------------------------

    ticket_similarities = (
        ticket_embeddings_norm @ query_embedding
    )

    # --------------------------------------
    # 4. Récupérer les tickets similaires
    # --------------------------------------

    top_indices = np.argsort(
        ticket_similarities
    )[::-1][:top_k_tickets]

    candidates = knowledge_base.iloc[
        top_indices
    ].copy()

    candidates["Ticket_Similarity"] = (
        ticket_similarities[top_indices]
    )

    # --------------------------------------
    # 5. Seuil de similarité ticket
    # --------------------------------------

    candidates = candidates[
        candidates["Ticket_Similarity"]
        >= min_ticket_similarity
    ].copy()

    # --------------------------------------
    # 6. Garder ACTIONABLE
    # --------------------------------------

    actionable = candidates[
        candidates["solution_type"] == "ACTIONABLE"
    ].copy()

    # --------------------------------------
    # 7. Similarité entre solutions
    # --------------------------------------

    solution_indices = actionable.index.tolist()

    if len(solution_indices) == 0:
        print("Aucune solution ACTIONABLE trouvée.")
        return None, None

    historical_solution_embeddings = (
        solution_embeddings[solution_indices]
    )

    solution_matrix = cosine_similarity(
        historical_solution_embeddings
    )

    # --------------------------------------
    # 8. Matrice de similarité
    # --------------------------------------

    solution_numbers = actionable["Number"].tolist()

    solution_similarity_df = pd.DataFrame(
        solution_matrix,
        index=solution_numbers,
        columns=solution_numbers
    )

    return actionable, solution_similarity_df
```

```python
# ==========================================
# SCORE DES CLUSTERS DE SOLUTIONS
# ==========================================

def score_solution_clusters(
    actionable,
    clusters
):
    """
    Calcule les caractéristiques de chaque cluster
    de solutions.

    Le rattachement des tickets se fait via la colonne
    'Number' et non via l'index pandas.
    """

    cluster_results = []

    for cluster_id in sorted(
        clusters["Cluster"].unique()
    ):

        # --------------------------------------
        # 1. Tickets du cluster
        # --------------------------------------

        cluster_tickets = clusters[
            clusters["Cluster"] == cluster_id
        ].index.tolist()

        # --------------------------------------
        # 2. Récupérer les tickets dans actionable
        #    grâce à leur numéro
        # --------------------------------------

        cluster_data = actionable[
            actionable["Number"].isin(cluster_tickets)
        ].copy()

        # Si aucun ticket correspondant
        if cluster_data.empty:
            continue

        # --------------------------------------
        # 3. Pertinence des tickets
        # --------------------------------------

        ticket_similarity_mean = (
            cluster_data["Ticket_Similarity"].mean()
        )

        ticket_similarity_max = (
            cluster_data["Ticket_Similarity"].max()
        )

        # --------------------------------------
        # 4. Cohérence des solutions
        # --------------------------------------

        valid_tickets = cluster_data["Number"].tolist()

        if len(valid_tickets) > 1:

            matrix = solution_similarity_matrix.loc[
                valid_tickets,
                valid_tickets
            ].values

            upper_values = matrix[
                np.triu_indices(
                    len(matrix),
                    k=1
                )
            ]

            solution_coherence = upper_values.mean()

        else:

            solution_coherence = 1.0

        # --------------------------------------
        # 5. Nombre de tickets
        # --------------------------------------

        number_of_tickets = len(cluster_data)

        # --------------------------------------
        # 6. Résultat
        # --------------------------------------

        cluster_results.append({

            "Cluster": cluster_id,

            "Number_of_Tickets":
                number_of_tickets,

            "Ticket_Similarity_Mean":
                ticket_similarity_mean,

            "Ticket_Similarity_Max":
                ticket_similarity_max,

            "Solution_Coherence":
                solution_coherence
        })

    return pd.DataFrame(cluster_results)
```

```python
# ==========================================
# CLASSIFICATION DES SOLUTIONS
# ==========================================

def classify_solution(solution):
    solution = str(solution).strip().lower()

    # Solutions administratives
    administrative_keywords = [
        "duplicate",
        "duplicated",
        "close the ticket",
        "ticket to avoid",
        "no response from user",
        "regional ticket opened",
        "new ticket has been raised",
        "a new ticket has been raised",
        "data provided",
        "cancelled",
        "cancelled by"
    ]

    for keyword in administrative_keywords:
        if keyword in solution:
            return "ADMINISTRATIVE"

    # Solutions génériques
    generic_solutions = {
        "resolved",
        "resolved.",
        "resolved,",
        "solved",
        "solved.",
        "done",
        "done.",
        "fixed",
        "fixed.",
        "closed",
        "closed.",
        "granted",
        "access granted",
        "access granted.",
        "necessary done",
        "necessary done.",
        "necessaey done.",
        "issue fixed",
        "issue solved",
        "problem solved"
    }

    if solution in generic_solutions:
        return "GENERIC"

    if solution.startswith("resolved inc"):
        return "GENERIC"

    if solution.startswith("fixed inc"):
        return "GENERIC"

    return "ACTIONABLE"
```

```python
# ==========================================
# CLASSEMENT DES CLUSTERS
# ==========================================

def rank_solution_clusters(
    cluster_scores,
    weight_relevance=0.5,
    weight_coherence=0.3,
    weight_support=0.2
):

    scores = cluster_scores.copy()

    # --------------------------------------
    # Cohérence
    # --------------------------------------
    # Pour un singleton, il n'y a pas de comparaison
    # avec une autre solution.
    # On considère donc la cohérence comme parfaite.
    scores["Solution_Coherence"] = (
        scores["Solution_Coherence"]
        .fillna(1.0)
    )

    # --------------------------------------
    # Support du cluster
    # --------------------------------------

    max_tickets = scores["Number_of_Tickets"].max()

    if max_tickets > 0:
        scores["Support"] = (
            scores["Number_of_Tickets"]
            / max_tickets
        )
    else:
        scores["Support"] = 0.0

    # --------------------------------------
    # Score final du cluster
    # --------------------------------------

    scores["Cluster_Final_Score"] = (

        weight_relevance
        * scores["Ticket_Similarity_Mean"]

        +

        weight_coherence
        * scores["Solution_Coherence"]

        +

        weight_support
        * scores["Support"]
    )

    # --------------------------------------
    # Classement
    # --------------------------------------

    scores = scores.sort_values(
        "Cluster_Final_Score",
        ascending=False
    ).reset_index(drop=True)

    return scores
```

```python
# ==========================================
# BACKTEST GLOBAL - TOP K TICKETS SIMILAIRES
# ==========================================

import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity


def run_backtest_top_k(
    actionable,
    model,
    k=3
):
    """
    Teste la recherche de tickets similaires
    sur chaque ticket historique.

    Pour chaque ticket :
    - le ticket lui-même est exclu
    - les autres tickets sont classés par similarité
    - on récupère les Top-K solutions
    """

    results = []

    # --------------------------------------
    # Embeddings de tous les tickets
    # --------------------------------------

    texts = actionable["text_for_embedding"].fillna("").tolist()

    embeddings = model.encode(
        texts,
        normalize_embeddings=True,
        show_progress_bar=True
    )

    # --------------------------------------
    # Matrice de similarité
    # --------------------------------------

    similarity_matrix = cosine_similarity(
        embeddings
    )

    numbers = actionable["Number"].tolist()

    # --------------------------------------
    # Backtest ticket par ticket
    # --------------------------------------

    for i, ticket_number in enumerate(numbers):

        true_solution = actionable.iloc[i]["Solution"]
        short_description = actionable.iloc[i]["Short Description"]

        similarities = similarity_matrix[i].copy()

        # Exclure le ticket lui-même
        similarities[i] = -1

        # Top K
        top_indices = np.argsort(
            similarities
        )[::-1][:k]

        for rank, idx in enumerate(top_indices, start=1):

            results.append({
                "Test_Ticket": ticket_number,
                "Short_Description": short_description,
                "True_Solution": true_solution,
                "Rank": rank,
                "Candidate_Ticket": numbers[idx],
                "Similarity": similarities[idx],
                "Candidate_Solution":
                    actionable.iloc[idx]["Solution"]
            })

    return pd.DataFrame(results)


backtest_results = run_backtest_top_k(
    actionable,
    model,
    k=3
)

print("===== BACKTEST TOP 3 =====")

print(
    backtest_results[
        [
            "Test_Ticket",
            "Rank",
            "Candidate_Ticket",
            "Similarity"
        ]
    ].to_string(index=False)
)
```

```python
# ==========================================
# EVALUATION : SIMILARITE TICKET vs SOLUTION
# ==========================================

def evaluate_solution_similarity(
    backtest_results,
    actionable,
    solution_similarity_matrix
):
    results = []

    for _, row in backtest_results.iterrows():

        test_ticket = row["Test_Ticket"]
        candidate_ticket = row["Candidate_Ticket"]

        # Vérifier que les deux tickets existent
        if (
            test_ticket not in solution_similarity_matrix.index
            or
            candidate_ticket not in solution_similarity_matrix.columns
        ):
            continue

        solution_similarity = (
            solution_similarity_matrix.loc[
                test_ticket,
                candidate_ticket
            ]
        )

        results.append({
            "Test_Ticket": test_ticket,
            "Rank": row["Rank"],
            "Candidate_Ticket": candidate_ticket,
            "Ticket_Similarity": row["Similarity"],
            "Solution_Similarity": solution_similarity
        })

    return pd.DataFrame(results)


solution_backtest = evaluate_solution_similarity(
    backtest_results,
    actionable,
    solution_similarity_matrix
)

print("===== BACKTEST TICKET + SOLUTION =====")

print(
    solution_backtest.to_string(
        index=False
    )
)
```

```python
# ==========================================
# BACKTEST END-TO-END — UN SEUL TICKET
# ==========================================

from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import AgglomerativeClustering
import numpy as np
import pandas as pd


def backtest_one_ticket(
    test_idx,
    data,
    model,
    top_k=5,
    cluster_threshold=0.75
):

    # ======================================
    # 1. TICKET DE TEST
    # ======================================

    test_ticket = data.iloc[test_idx]

    test_number = test_ticket["Number"]
    true_solution = test_ticket["Solution"]

    print("=" * 60)
    print("TICKET DE TEST")
    print("=" * 60)

    print("Number :", test_number)
    print("Description :", test_ticket["Short Description"])

    print("\nVRAIE SOLUTION (cachée au système) :")
    print(true_solution)


    # ======================================
    # 2. EMBEDDING DU TICKET DE TEST
    # ======================================

    query_embedding = model.encode(
        [test_ticket["text_for_embedding"]],
        normalize_embeddings=True
    )[0]


    # ======================================
    # 3. EMBEDDINGS DES AUTRES TICKETS
    # ======================================

    candidate_data = data[
        data["Number"] != test_number
    ].copy()

    candidate_texts = (
        candidate_data["text_for_embedding"]
        .fillna("")
        .tolist()
    )

    candidate_embeddings = model.encode(
        candidate_texts,
        normalize_embeddings=True
    )


    # ======================================
    # 4. SIMILARITÉ TICKET ↔ TICKET
    # ======================================

    ticket_similarities = cosine_similarity(
        [query_embedding],
        candidate_embeddings
    )[0]

    candidate_data["Ticket_Similarity"] = (
        ticket_similarities
    )


    # ======================================
    # 5. TOP-K TICKETS
    # ======================================

    candidate_data = candidate_data.sort_values(
        "Ticket_Similarity",
        ascending=False
    ).head(top_k).copy()


    print("\n" + "=" * 60)
    print("TOP TICKETS SIMILAIRES")
    print("=" * 60)

    print(
        candidate_data[
            [
                "Number",
                "Ticket_Similarity",
                "Short Description",
                "Solution"
            ]
        ].to_string(index=False)
    )


    # ======================================
    # 6. EMBEDDINGS DES SOLUTIONS
    # ======================================

    solution_texts = (
        candidate_data["Solution"]
        .fillna("")
        .tolist()
    )

    candidate_solution_embeddings = model.encode(
        solution_texts,
        normalize_embeddings=True
    )


    # ======================================
    # 7. MATRICE DE SIMILARITÉ DES SOLUTIONS
    # ======================================

    solution_matrix = cosine_similarity(
        candidate_solution_embeddings
    )


    solution_similarity_matrix_test = pd.DataFrame(
        solution_matrix,
        index=candidate_data["Number"],
        columns=candidate_data["Number"]
    )


    # ======================================
    # 8. CLUSTERING DES SOLUTIONS
    # ======================================

    distance_matrix = 1 - solution_matrix

    clustering = AgglomerativeClustering(
        n_clusters=None,
        metric="precomputed",
        linkage="average",
        distance_threshold=1 - cluster_threshold
    )

    labels = clustering.fit_predict(
        distance_matrix
    )

    candidate_data["Cluster"] = labels


    print("\n" + "=" * 60)
    print("CLUSTERS")
    print("=" * 60)

    for cluster_id in sorted(
        candidate_data["Cluster"].unique()
    ):

        cluster_rows = candidate_data[
            candidate_data["Cluster"] == cluster_id
        ]

        print(f"\nCluster {cluster_id}")

        for _, row in cluster_rows.iterrows():

            print(
                f"  {row['Number']} | "
                f"Ticket similarity = "
                f"{row['Ticket_Similarity']:.4f}"
            )


    # ======================================
    # 9. SCORE DES CLUSTERS
    # ======================================

    cluster_results = []

    for cluster_id in sorted(
        candidate_data["Cluster"].unique()
    ):

        cluster_data = candidate_data[
            candidate_data["Cluster"] == cluster_id
        ]

        numbers = cluster_data[
            "Number"
        ].tolist()


        # Pertinence moyenne
        ticket_mean = cluster_data[
            "Ticket_Similarity"
        ].mean()


        # Pertinence maximale
        ticket_max = cluster_data[
            "Ticket_Similarity"
        ].max()


        # Cohérence des solutions
        if len(numbers) > 1:

            matrix = solution_similarity_matrix_test.loc[
                numbers,
                numbers
            ].values

            upper_values = matrix[
                np.triu_indices(
                    len(numbers),
                    k=1
                )
            ]

            coherence = upper_values.mean()

        else:

            coherence = 1.0


        # Nombre de preuves
        support = len(cluster_data)


        cluster_results.append({

            "Cluster": cluster_id,

            "Number_of_Tickets":
                support,

            "Ticket_Similarity_Mean":
                ticket_mean,

            "Ticket_Similarity_Max":
                ticket_max,

            "Solution_Coherence":
                coherence
        })


    cluster_scores_test = pd.DataFrame(
        cluster_results
    )


    # ======================================
    # 10. SUPPORT NORMALISÉ
    # ======================================

    max_support = (
        cluster_scores_test[
            "Number_of_Tickets"
        ].max()
    )

    cluster_scores_test["Support"] = (
        cluster_scores_test[
            "Number_of_Tickets"
        ] / max_support
    )


    # ======================================
    # 11. SCORE FINAL DU CLUSTER
    # ======================================

    cluster_scores_test[
        "Cluster_Final_Score"
    ] = (

        0.5 *
        cluster_scores_test[
            "Ticket_Similarity_Mean"
        ]

        +

        0.3 *
        cluster_scores_test[
            "Solution_Coherence"
        ]

        +

        0.2 *
        cluster_scores_test[
            "Support"
        ]
    )


    cluster_scores_test = (
        cluster_scores_test
        .sort_values(
            "Cluster_Final_Score",
            ascending=False
        )
        .reset_index(drop=True)
    )


    print("\n" + "=" * 60)
    print("CLASSEMENT DES CLUSTERS")
    print("=" * 60)

    print(
        cluster_scores_test.to_string(
            index=False
        )
    )


    # ======================================
    # 12. CLUSTER GAGNANT
    # ======================================

    best_cluster = int(
        cluster_scores_test.iloc[0]["Cluster"]
    )

    winning_cluster_data = candidate_data[
        candidate_data["Cluster"] == best_cluster
    ].copy()


    # ======================================
    # 13. SOLUTION REPRESENTATIVE
    # ======================================

    winning_numbers = (
        winning_cluster_data["Number"]
        .tolist()
    )

    if len(winning_numbers) == 1:

        representative_number = (
            winning_numbers[0]
        )

        internal_similarity = 1.0

    else:

        matrix = solution_similarity_matrix_test.loc[
            winning_numbers,
            winning_numbers
        ]

        mean_similarities = {}

        for number in winning_numbers:

            others = [
                n for n in winning_numbers
                if n != number
            ]

            mean_similarities[number] = (
                matrix.loc[
                    number,
                    others
                ].mean()
            )

        representative_number = max(
            mean_similarities,
            key=mean_similarities.get
        )

        internal_similarity = (
            mean_similarities[
                representative_number
            ]
        )


    representative_row = winning_cluster_data[
        winning_cluster_data["Number"]
        == representative_number
    ].iloc[0]


    recommended_solution = (
        representative_row["Solution"]
    )


    # ======================================
    # 14. RÉSULTAT FINAL
    # ======================================

    print("\n" + "=" * 60)
    print("RÉSULTAT DU BACKTEST")
    print("=" * 60)

    print("\nTicket de test :")
    print(test_number)

    print("\nCluster gagnant :")
    print(best_cluster)

    print("\nTicket représentant :")
    print(representative_number)

    print("\nSimilarité ticket du représentant :")
    print(
        round(
            representative_row[
                "Ticket_Similarity"
            ],
            4
        )
    )

    print("\nCohérence interne :")
    print(
        round(
            internal_similarity,
            4
        )
    )

    print("\nSOLUTION RECOMMANDÉE :")
    print(recommended_solution)

    print("\nVRAIE SOLUTION :")
    print(true_solution)


    return {
        "Test_Ticket": test_number,
        "True_Solution": true_solution,
        "Best_Cluster": best_cluster,
        "Representative_Ticket": representative_number,
        "Recommended_Solution": recommended_solution,
        "Cluster_Score":
            cluster_scores_test.iloc[0][
                "Cluster_Final_Score"
            ],
        "Cluster_Scores":
            cluster_scores_test,
        "Candidates":
            candidate_data,
        "Solution_Matrix":
            solution_similarity_matrix_test
    }
```

```python
# ==========================================
# BACKTEST END-TO-END GLOBAL
# ==========================================

def run_end_to_end_backtest(
    evaluation_data,
    model,
    top_k=5,
    cluster_threshold=0.75
):
    """
    Backtest complet du pipeline.

    Pour chaque ticket :
    - le ticket est considéré comme nouveau
    - sa vraie solution est cachée
    - les autres tickets servent de base historique
    - recherche des tickets similaires
    - filtrage ACTIONABLE
    - clustering des solutions
    - scoring des clusters
    - sélection d'une solution représentative
    - comparaison avec la vraie solution
    """

    results = []

    # Embeddings des tickets
    ticket_texts = (
        evaluation_data["text_for_embedding"]
        .fillna("")
        .tolist()
    )

    ticket_embeddings = model.encode(
        ticket_texts,
        normalize_embeddings=True,
        show_progress_bar=False
    )

    # Embeddings des solutions
    solution_texts = (
        evaluation_data["Solution"]
        .fillna("")
        .tolist()
    )

    solution_embeddings_all = model.encode(
        solution_texts,
        normalize_embeddings=True,
        show_progress_bar=False
    )

    numbers = evaluation_data["Number"].tolist()

    # ======================================
    # TICKET PAR TICKET
    # ======================================

    for test_idx, test_number in enumerate(numbers):

        print(
            f"Backtest {test_idx + 1}/{len(numbers)} : "
            f"{test_number}"
        )

        test_row = evaluation_data.iloc[test_idx]

        true_solution = test_row["Solution"]

        # ----------------------------------
        # 1. Embedding du ticket de test
        # ----------------------------------

        query_embedding = ticket_embeddings[test_idx]

        # ----------------------------------
        # 2. Similarité avec tous les tickets
        # ----------------------------------

        ticket_similarities = (
            ticket_embeddings @ query_embedding
        )

        # ----------------------------------
        # 3. Exclure le ticket lui-même
        # ----------------------------------

        ticket_similarities[test_idx] = -1

        # ----------------------------------
        # 4. Top K
        # ----------------------------------

        top_indices = np.argsort(
            ticket_similarities
        )[::-1][:top_k]

        candidates = evaluation_data.iloc[
            top_indices
        ].copy()

        candidates["Ticket_Similarity"] = (
            ticket_similarities[top_indices]
        )

        # ----------------------------------
        # 5. ACTIONABLE uniquement
        # ----------------------------


        actionable_test = candidates[
            candidates["solution_type"] == "ACTIONABLE"
        ].copy()

        if actionable_test.empty:
            continue

        # ----------------------------------
        # 6. Embeddings des solutions
        # ----------------------------------

        actionable_indices = actionable_test.index.tolist()

        actionable_solution_embeddings = (
            solution_embeddings_all[
                actionable_indices
            ]
        )

        # ----------------------------------
        # 7. Similarité entre solutions
        # ----------------------------------

        solution_similarity_values = cosine_similarity(
            actionable_solution_embeddings
        )

        solution_similarity_matrix_test = pd.DataFrame(
            solution_similarity_values,
            index=actionable_test["Number"],
            columns=actionable_test["Number"]
        )

        # ----------------------------------
        # 8. Clustering
        # ----------------------------------

        clusters_test = cluster_similar_solutions(
            solution_similarity_matrix_test,
            threshold=cluster_threshold
        )

        # ----------------------------------
        # 9. Score des clusters
        # ----------------------------------

        cluster_scores_test = score_solution_clusters(
            actionable_test,
            clusters_test,
            solution_similarity_matrix_test
        )

        # ----------------------------------
        # 10. Classement des clusters
        # ----------------------------------

        ranked_clusters_test = rank_solution_clusters(
            cluster_scores_test
        )

        if ranked_clusters_test.empty:
            continue

        # ----------------------------------
        # 11. Cluster gagnant
        # ----------------------------------

        best_cluster = int(
            ranked_clusters_test.iloc[0]["Cluster"]
        )

        best_cluster_numbers = clusters_test[
            clusters_test["Cluster"] == best_cluster
        ]["Number"].tolist()

        # ----------------------------------
        # 12. Trouver la solution représentative
        # ----------------------------------

        cluster_data = actionable_test[
            actionable_test["Number"].isin(
                best_cluster_numbers
            )
        ].copy()

        if len(best_cluster_numbers) == 1:

            representative_row = (
                cluster_data.iloc[0]
            )

            internal_similarity = 1.0

        else:

            matrix = solution_similarity_matrix_test.loc[
                best_cluster_numbers,
                best_cluster_numbers
            ]

            mean_similarities = {}

            for ticket_number in best_cluster_numbers:

                others = [
                    t
                    for t in best_cluster_numbers
                    if t != ticket_number
                ]

                mean_similarities[ticket_number] = (
                    matrix.loc[
                        ticket_number,
                        others
                    ].mean()
                )

            representative_ticket = max(
                mean_similarities,
                key=mean_similarities.get
            )

            representative_row = cluster_data[
                cluster_data["Number"]
                == representative_ticket
            ].iloc[0]

            internal_similarity = (
                mean_similarities[
                    representative_ticket
                ]
            )

        # ----------------------------------
        # 13. Solution recommandée
        # ----------------------------------

        recommended_solution = (
            representative_row["Solution"]
        )

        representative_ticket = (
            representative_row["Number"]
        )

        ticket_similarity = (
            representative_row["Ticket_Similarity"]
        )

        # ----------------------------------
        # 14. Comparaison avec vraie solution
        # ----------------------------------

        true_solution_embedding = model.encode(
            [true_solution],
            normalize_embeddings=True
        )[0]

        recommended_solution_embedding = (
            model.encode(
                [recommended_solution],
                normalize_embeddings=True
            )[0]
        )

        solution_similarity_to_true = (
            true_solution_embedding
            @ recommended_solution_embedding
        )

        # ----------------------------------
        # 15. Résultat
        # ----------------------------------

        results.append({

            "Test_Ticket":
                test_number,

            "True_Solution":
                true_solution,

            "Best_Cluster":
                best_cluster,

            "Representative_Ticket":
                representative_ticket,

            "Ticket_Similarity":
                ticket_similarity,

            "Internal_Similarity":
                internal_similarity,

            "Cluster_Score":
                ranked_clusters_test.iloc[0][
                    "Cluster_Final_Score"
                ],

            "Recommended_Solution":
                recommended_solution,

            "Solution_Similarity":
                solution_similarity_to_true,

            "Number_of_Supporting_Tickets":
                len(best_cluster_numbers)
        })

    return pd.DataFrame(results)
```

```python
# ==========================================
# CONSOLIDATION DES ÉVIDENCES
# ==========================================

def build_recommendation_evidence(
    ranked_clusters,
    clusters,
    actionable,
    representative_solutions
):
    """
    Construit les preuves historiques utilisées
    pour générer la recommandation finale.

    Aucune nouvelle solution n'est générée ici.
    """

    # --------------------------------------
    # 1. Cluster gagnant
    # --------------------------------------

    best_cluster = int(
        ranked_clusters.iloc[0]["Cluster"]
    )

    best_cluster_row = ranked_clusters.iloc[0]

    # --------------------------------------
    # 2. Tickets appartenant au cluster
    # --------------------------------------

    cluster_numbers = clusters.loc[
        clusters["Cluster"] == best_cluster,
        "Number"
    ].tolist()

    # --------------------------------------
    # 3. Récupérer les données historiques
    # --------------------------------------

    cluster_data = actionable[
        actionable["Number"].isin(cluster_numbers)
    ].copy()

    # Trier par pertinence avec le nouveau ticket
    cluster_data = cluster_data.sort_values(
        "Ticket_Similarity",
        ascending=False
    )

    # --------------------------------------
    # 4. Construire les preuves
    # --------------------------------------

    supporting_tickets = []

    for _, row in cluster_data.iterrows():

        supporting_tickets.append({

            "Number":
                row["Number"],

            "Short_Description":
                row["Short Description"],

            "Ticket_Similarity":
                float(row["Ticket_Similarity"]),

            "Solution":
                row["Solution"]
        })

    # --------------------------------------
    # 5. Solution représentative
    # --------------------------------------

    representative = representative_solutions[
        representative_solutions["Cluster"]
        == best_cluster
    ]

    if not representative.empty:

        representative = representative.iloc[0]

        representative_ticket = (
            representative["Number"]
        )

        representative_solution = (
            representative["Solution"]
        )

    else:

        representative_ticket = None
        representative_solution = None

    # --------------------------------------
    # 6. Objet final d'évidence
    # --------------------------------------

    evidence = {

        "Best_Cluster":
            best_cluster,

        "Cluster_Score":
            float(
                best_cluster_row[
                    "Cluster_Final_Score"
                ]
            ),

        "Solution_Coherence":
            float(
                best_cluster_row[
                    "Solution_Coherence"
                ]
            ),

        "Number_of_Supporting_Tickets":
            len(supporting_tickets),

        "Supporting_Tickets":
            supporting_tickets,

        "Representative_Ticket":
            representative_ticket,

        "Representative_Solution":
            representative_solution
    }

    return evidence
```

```python
# ==========================================
# GÉNÉRATION DE LA RECOMMANDATION
# À PARTIR DES ÉVIDENCES HISTORIQUES
# ==========================================

def generate_recommendation_from_evidence(
    recommendation_evidence
):
    """
    Génère une recommandation naturelle à partir
    des solutions historiques du cluster gagnant.

    Important :
    aucune nouvelle solution n'est inventée.
    """

    supporting_tickets = (
        recommendation_evidence[
            "Supporting_Tickets"
        ]
    )

    if not supporting_tickets:
        return (
            "No reliable historical solution was "
            "found for this incident."
        )

    # --------------------------------------
    # Récupérer les solutions historiques
    # --------------------------------------

    solutions = [
        str(ticket["Solution"]).strip()
        for ticket in supporting_tickets
        if str(ticket["Solution"]).strip()
    ]

    # --------------------------------------
    # Détecter les actions présentes
    # dans les solutions historiques
    # --------------------------------------

    actions = []

    for solution in solutions:

        solution_lower = solution.lower()

        if (
            "printer setup" in solution_lower
            or "printer configuration" in solution_lower
        ):
            actions.append(
                "review and update the printer setup"
            )

        if "printer driver" in solution_lower:
            actions.append(
                "update the printer drivers"
            )

        if (
            "printer has been fixed" in solution_lower
            or "working now" in solution_lower
            or "issue resolved" in solution_lower
        ):
            actions.append(
                "verify that the printer is working correctly"
            )

    # Supprimer les doublons
    actions = list(dict.fromkeys(actions))

    # --------------------------------------
    # Construire la recommandation
    # --------------------------------------

    if not actions:

        return (
            "Based on similar historical incidents, "
            "review the solutions from the supporting "
            "historical tickets."
        )

    recommendation_parts = []

    if any(
        "printer setup" in action
        for action in actions
    ):
        recommendation_parts.append(
            "review and update the printer setup"
        )

    if any(
        "printer drivers" in action
        for action in actions
    ):
        recommendation_parts.append(
            "update the printer drivers"
        )

    # Première partie : actions
    if len(recommendation_parts) == 2:

        action_sentence = (
            f"Based on similar historical incidents, "
            f"review and update the printer setup "
            f"and update the printer drivers as required."
        )

    elif len(recommendation_parts) == 1:

        action_sentence = (
            "Based on similar historical incidents, "
            f"{recommendation_parts[0]} as required."
        )

    else:

        action_sentence = (
            "Based on similar historical incidents, "
            "review the applicable printer configuration."
        )

    # --------------------------------------
    # Vérification finale
    # --------------------------------------

    if any(
        "verify" in action
        for action in actions
    ):

        action_sentence += (
            " Then verify that the printer is "
            "working correctly."
        )

    return action_sentence
```

```python
# ==========================================
# BACKTEST END-TO-END DE LA RECOMMANDATION
# ==========================================

def run_end_to_end_backtest(
    evaluation_data,
    model,
    top_k=5,
    cluster_threshold=0.75
):
    """
    Évalue tout le pipeline sur chaque ticket historique.

    Pour chaque ticket :
        1. Le ticket est retiré de la recherche
        2. Recherche des tickets similaires
        3. Conservation des solutions ACTIONABLE
        4. Clustering des solutions
        5. Sélection du cluster gagnant
        6. Construction des évidences
        7. Génération de la recommandation
    """

    results = []

    numbers = evaluation_data["Number"].tolist()

    for test_idx, test_number in enumerate(numbers):

        print(
            f"Backtest {test_idx + 1}/{len(numbers)} : "
            f"{test_number}"
        )

        # ======================================
        # 1. Ticket de test
        # ======================================

        test_ticket = evaluation_data.iloc[test_idx]

        true_solution = test_ticket["Solution"]

        # ======================================
        # 2. Base historique
        # ======================================

        historical_data = evaluation_data[
            evaluation_data["Number"] != test_number
        ].copy()

        historical_data = historical_data.reset_index(
            drop=True
        )

        if len(historical_data) == 0:
            continue

        # ======================================
        # 3. Embeddings des tickets historiques
        # ======================================

        historical_embeddings = model.encode(
            historical_data[
                "text_for_embedding"
            ].fillna("").tolist(),
            normalize_embeddings=True,
            show_progress_bar=False
        )

        # ======================================
        # 4. Embedding du ticket de test
        # ======================================

        query_embedding = model.encode(
            [test_ticket["text_for_embedding"]],
            normalize_embeddings=True
        )[0]

        # ======================================
        # 5. Similarité ticket ↔ historique
        # ======================================

        similarities = cosine_similarity(
            [query_embedding],
            historical_embeddings
        )[0]

        historical_data[
            "Ticket_Similarity"
        ] = similarities

        # ======================================
        # 6. Top K
        # ======================================

        top_indices = np.argsort(
            similarities
        )[::-1][:top_k]

        candidates = historical_data.iloc[
            top_indices
        ].copy()

        # ======================================
        # 7. ACTIONABLE uniquement
        # ======================================

        candidates = candidates[
            candidates["solution_type"]
            == "ACTIONABLE"
        ].copy()

        if candidates.empty:
            continue

        candidates = candidates.sort_values(
            "Ticket_Similarity",
            ascending=False
        )

        candidates = candidates.reset_index(
            drop=True
        )

        # ======================================
        # 8. Embeddings des solutions
        # ======================================

        candidate_solution_embeddings = model.encode(
            candidates["Solution"]
            .fillna("")
            .tolist(),
            normalize_embeddings=True,
            show_progress_bar=False
        )

        # ======================================
        # 9. Matrice de similarité des solutions
        # ======================================

        solution_matrix = cosine_similarity(
            candidate_solution_embeddings
        )

        solution_similarity_matrix_local = pd.DataFrame(
            solution_matrix,
            index=candidates["Number"],
            columns=candidates["Number"]
        )

        # ======================================
        # 10. Clustering
        # ======================================

        local_clusters = cluster_similar_solutions(
            solution_similarity_matrix_local,
            threshold=cluster_threshold
        )

        # ======================================
        # 11. Statistiques des clusters
        # ======================================

        local_cluster_scores = score_solution_clusters(
            candidates,
            local_clusters,
            solution_similarity_matrix_local
        )

        # ======================================
        # 12. Classement des clusters
        # ======================================

        local_ranked_clusters = rank_solution_clusters(
            local_cluster_scores
        )

        # ======================================
        # 13. Solutions représentatives
        # ======================================

        local_representatives = []

        for cluster_id in sorted(
            local_clusters["Cluster"].unique()
        ):

            cluster_numbers = local_clusters.loc[
                local_clusters["Cluster"] == cluster_id,
                "Number"
            ].tolist()

            cluster_data = candidates[
                candidates["Number"].isin(
                    cluster_numbers
                )
            ]

            if len(cluster_numbers) == 1:

                row = cluster_data.iloc[0]

                local_representatives.append({

                    "Cluster": cluster_id,

                    "Number":
                        row["Number"],

                    "Solution":
                        row["Solution"],

                    "Ticket_Similarity":
                        row["Ticket_Similarity"],

                    "Internal_Similarity":
                        1.0
                })

            else:

                matrix = (
                    solution_similarity_matrix_local
                    .loc[
                        cluster_numbers,
                        cluster_numbers
                    ]
                )

                mean_similarities = {}

                for ticket_number in cluster_numbers:

                    others = [
                        t
                        for t in cluster_numbers
                        if t != ticket_number
                    ]

                    mean_similarities[
                        ticket_number
                    ] = matrix.loc[
                        ticket_number,
                        others
                    ].mean()

                representative_ticket = max(
                    mean_similarities,
                    key=mean_similarities.get
                )

                row = cluster_data[
                    cluster_data["Number"]
                    == representative_ticket
                ].iloc[0]

                local_representatives.append({

                    "Cluster":
                        cluster_id,

                    "Number":
                        representative_ticket,

                    "Solution":
                        row["Solution"],

                    "Ticket_Similarity":
                        row["Ticket_Similarity"],

                    "Internal_Similarity":
                        mean_similarities[
                            representative_ticket
                        ]
                })

        local_representatives = pd.DataFrame(
            local_representatives
        )

        # ======================================
        # 14. Cluster gagnant
        # ======================================

        best_cluster = int(
            local_ranked_clusters.iloc[0]["Cluster"]
        )

        # ======================================
        # 15. Construire les évidences
        # ======================================

        local_evidence = (
            build_recommendation_evidence(
                local_ranked_clusters,
                local_clusters,
                candidates,
                local_representatives
            )
        )

        # ======================================
        # 16. Générer recommandation
        # ======================================

        recommendation = (
            generate_recommendation_from_evidence(
                local_evidence
            )
        )

        # ======================================
        # 17. Représentant
        # ======================================

        representative = local_representatives[
            local_representatives["Cluster"]
            == best_cluster
        ].iloc[0]

        # ======================================
        # 18. Similarité solution
        # ======================================

        representative_ticket = (
            representative["Number"]
        )

        solution_similarity = (
            cosine_similarity(
                model.encode(
                    [true_solution],
                    normalize_embeddings=True
                ),
                model.encode(
                    [representative["Solution"]],
                    normalize_embeddings=True
                )
            )[0][0]
        )

        # ======================================
        # 19. Résultat
        # ======================================

        best_cluster_row = (
            local_ranked_clusters.iloc[0]
        )

        results.append({

            "Test_Ticket":
                test_number,

            "True_Solution":
                true_solution,

            "Best_Cluster":
                best_cluster,

            "Representative_Ticket":
                representative_ticket,

            "Ticket_Similarity":
                representative[
                    "Ticket_Similarity"
                ],

            "Internal_Similarity":
                representative[
                    "Internal_Similarity"
                ],

            "Cluster_Score":
                best_cluster_row[
                    "Cluster_Final_Score"
                ],

            "Solution_Similarity":
                solution_similarity,

            "Recommendation":
                recommendation
        })

    return pd.DataFrame(results)
```

```python
# ==========================================
# DIAGNOSTIC DU BACKTEST
# COMPARAISON PIPELINE vs MEILLEURE SOLUTION
# ==========================================

diagnostic_results = []

# Embeddings des solutions
solution_embeddings_eval = model.encode(
    evaluation_data["Solution"].fillna("").tolist(),
    normalize_embeddings=True,
    show_progress_bar=False
)

solution_matrix_eval = cosine_similarity(
    solution_embeddings_eval
)

numbers = evaluation_data["Number"].tolist()


for _, result in end_to_end_results.iterrows():

    test_ticket = result["Test_Ticket"]
    representative = result["Representative_Ticket"]

    # --------------------------------------
    # Position du ticket de test
    # --------------------------------------

    test_idx = numbers.index(test_ticket)

    # --------------------------------------
    # Similarités solution avec tous les
    # autres tickets
    # --------------------------------------

    solution_similarities = (
        solution_matrix_eval[test_idx].copy()
    )

    # Exclure le ticket lui-même
    solution_similarities[test_idx] = -1

    # Meilleure solution historique
    best_idx = np.argmax(
        solution_similarities
    )

    best_ticket = numbers[best_idx]
    best_solution_similarity = (
        solution_similarities[best_idx]
    )

    # --------------------------------------
    # Résultat du pipeline
    # --------------------------------------

    pipeline_solution_similarity = (
        result["Solution_Similarity"]
    )

    # --------------------------------------
    # Ecart avec la meilleure solution
    # --------------------------------------

    gap = (
        best_solution_similarity
        -
        pipeline_solution_similarity
    )

    diagnostic_results.append({

        "Test_Ticket":
            test_ticket,

        "Pipeline_Representative":
            representative,

        "Pipeline_Solution_Similarity":
            pipeline_solution_similarity,

        "Best_Historical_Ticket":
            best_ticket,

        "Best_Solution_Similarity":
            best_solution_similarity,

        "Gap":
            gap
    })


diagnostic_results = pd.DataFrame(
    diagnostic_results
)

print("=" * 70)
print("DIAGNOSTIC : PIPELINE VS MEILLEURE SOLUTION HISTORIQUE")
print("=" * 70)

print(
    diagnostic_results
    .sort_values(
        "Gap",
        ascending=False
    )
    .round(4)
    .to_string(index=False)
)
```

```python
# ==========================================
# ANALYSE DES CLUSTERS CANDIDATS
# ==========================================

def analyze_backtest_clusters(
    evaluation_data,
    end_to_end_results,
    model,
    k=5
):

    diagnostic = []

    numbers = evaluation_data["Number"].tolist()

    # --------------------------------------
    # Embeddings des tickets
    # --------------------------------------

    ticket_embeddings = model.encode(
        evaluation_data["text_for_embedding"]
        .fillna("")
        .tolist(),
        normalize_embeddings=True,
        show_progress_bar=False
    )

    ticket_similarity_matrix = cosine_similarity(
        ticket_embeddings
    )

    # --------------------------------------
    # Embeddings des solutions
    # --------------------------------------

    solution_embeddings = model.encode(
        evaluation_data["Solution"]
        .fillna("")
        .tolist(),
        normalize_embeddings=True,
        show_progress_bar=False
    )

    solution_similarity_matrix_eval = cosine_similarity(
        solution_embeddings
    )

    # --------------------------------------
    # Analyse ticket par ticket
    # --------------------------------------

    for _, result in end_to_end_results.iterrows():

        test_ticket = result["Test_Ticket"]

        test_idx = numbers.index(test_ticket)

        # ----------------------------------
        # Top K tickets similaires
        # ----------------------------------

        similarities = (
            ticket_similarity_matrix[test_idx].copy()
        )

        # Exclure le ticket lui-même
        similarities[test_idx] = -1

        top_indices = np.argsort(
            similarities
        )[::-1][:k]

        # ----------------------------------
        # Construire les informations
        # ----------------------------------

        for idx in top_indices:

            candidate_ticket = numbers[idx]

            # Trouver le cluster du candidat
            cluster_row = clusters[
                clusters["Number"] == candidate_ticket
            ]

            if cluster_row.empty:
                continue

            cluster_id = int(
                cluster_row.iloc[0]["Cluster"]
            )

            # ----------------------------------
            # Statistiques du cluster
            # ----------------------------------

            cluster_numbers = clusters.loc[
                clusters["Cluster"] == cluster_id,
                "Number"
            ].tolist()

            cluster_data = evaluation_data[
                evaluation_data["Number"].isin(
                    cluster_numbers
                )
            ]

            # Similarité moyenne des tickets
            ticket_mean = cluster_data[
                "Ticket_Similarity"
            ].mean()

            # Similarité maximale
            ticket_max = cluster_data[
                "Ticket_Similarity"
            ].max()

            # ----------------------------------
            # Cohérence des solutions
            # ----------------------------------

            if len(cluster_numbers) > 1:

                indices_cluster = [
                    numbers.index(t)
                    for t in cluster_numbers
                ]

                matrix = solution_similarity_matrix_eval[
                    np.ix_(
                        indices_cluster,
                        indices_cluster
                    )
                ]

                upper_values = matrix[
                    np.triu_indices(
                        len(matrix),
                        k=1
                    )
                ]

                solution_coherence = (
                    upper_values.mean()
                )

            else:

                solution_coherence = 1.0

            # ----------------------------------
            # Support
            # ----------------------------------

            support = len(cluster_data)

            max_support = len(evaluation_data)

            support_normalized = (
                support / max_support
            )

            # ----------------------------------
            # Score du cluster
            # ----------------------------------

            cluster_score = (
                0.5 * ticket_mean
                +
                0.3 * solution_coherence
                +
                0.2 * support_normalized
            )

            # ----------------------------------
            # Similarité avec la vraie solution
            #
            # UNIQUEMENT pour le diagnostic
            # ----------------------------------

            true_solution_similarity = (
                solution_similarity_matrix_eval[
                    test_idx
                ]
            )

            cluster_solution_similarities = (
                true_solution_similarity[
                    [
                        numbers.index(t)
                        for t in cluster_numbers
                    ]
                ]
            )

            cluster_true_solution_mean = (
                cluster_solution_similarities.mean()
            )

            cluster_true_solution_max = (
                cluster_solution_similarities.max()
            )

            # ----------------------------------
            # Résultat
            # ----------------------------------

            diagnostic.append({

                "Test_Ticket":
                    test_ticket,

                "Cluster":
                    cluster_id,

                "Candidate_From_Cluster":
                    candidate_ticket,

                "Ticket_Similarity":
                    similarities[idx],

                "Ticket_Similarity_Mean":
                    ticket_mean,

                "Ticket_Similarity_Max":
                    ticket_max,

                "Solution_Coherence":
                    solution_coherence,

                "Support":
                    support,

                "Cluster_Score":
                    cluster_score,

                # Oracle / diagnostic seulement
                "True_Solution_Similarity_Mean":
                    cluster_true_solution_mean,

                "True_Solution_Similarity_Max":
                    cluster_true_solution_max
            })

    return pd.DataFrame(diagnostic)


cluster_diagnostic = analyze_backtest_clusters(
    evaluation_data,
    end_to_end_results,
    model,
    k=5
)

print("=" * 80)
print("ANALYSE DES CLUSTERS CANDIDATS")
print("=" * 80)

print(
    cluster_diagnostic
    .sort_values(
        ["Test_Ticket", "Cluster_Score"],
        ascending=[True, False]
    )
    .round(4)
    .to_string(index=False)
)
```
